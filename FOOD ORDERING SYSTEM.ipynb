```python
class MenuItem:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class Restaurant:
    def show_menu(self):
        print("\n1. Pizza - 200")
        print("2. Burger - 100")

class Order:
    def __init__(self):
        self.price = 0
        self.status = "Not Ordered"

    def bill(self):
        print("Total Bill:", self.price)

    def track(self):
        print("Order Status:", self.status)

class Customer:
    def __init__(self, name):
        self.name = name

r = Restaurant()
o = Order()
c = Customer(input("Enter name: "))

while True:
    print("\n1. Browse Menu")
    print("2. Place Order")
    print("3. Calculate Bill")
    print("4. Track Order")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        r.show_menu()

    elif choice == 2:
        r.show_menu()
        n = int(input("Select food: "))

        if n == 1:
            o.price += 200
            o.status = "Order Placed"
        elif n == 2:
            o.price += 100
            o.status = "Order Placed"
        else:
            print("Invalid choice")

    elif choice == 3:
        o.bill()

    elif choice == 4:
        o.track()

    elif choice == 5:
        print("Thank you", c.name)
        break

    else:
        print("Invalid choice")
```
